```python

import requests
from bs4 import BeautifulSoup
import csv
import json
import os

def scrape_motivational_quotes(html_content, output_folder="BrainyQuote_Data"):

    print("Starting Motivational Quotes Extraction...")

    try:
        # Parse HTML content
        soup = BeautifulSoup(html_content, "html.parser")

        quotes_data = []

        # Find all quote containers
        quotes = soup.find_all("div", class_="quote")

        for item in quotes:

            quote = item.find("a", title="view quote")
            author = item.find("a", title="view author")

            if quote and author:

                quote_text = quote.get_text(strip=True)
                author_name = author.get_text(strip=True)

                quotes_data.append({
                    "Quote": quote_text,
                    "Author": author_name,
                    "Category": "Motivational"
                })

        # Check extracted data
        if not quotes_data:
            print("No quotes found. Check the HTML structure.")
            return

        # Create output directory
        os.makedirs(output_folder, exist_ok=True)

        # Save data in CSV format
        csv_path = os.path.join(
            output_folder,
            "motivational_quotes.csv"
        )

        try:
            with open(csv_path, "w", newline="", encoding="utf-8") as file:

                writer = csv.DictWriter(
                    file,
                    fieldnames=["Quote", "Author", "Category"]
                )

                writer.writeheader()
                writer.writerows(quotes_data)

            print("CSV file saved successfully!")

        except IOError as e:
            print("CSV writing error:", e)

        # Save data in JSON format
        json_path = os.path.join(
            output_folder,
            "motivational_quotes.json"
        )

        try:
            with open(json_path, "w", encoding="utf-8") as file:

                json.dump(
                    quotes_data,
                    file,
                    indent=4,
                    ensure_ascii=False
                )

            print("JSON file saved successfully!")

        except IOError as e:
            print("JSON writing error:", e)

        # Display extracted records
        print("\nExtracted Motivational Quotes:")
        print("  ")

        for index, item in enumerate(quotes_data, start=1):

            print(f"\nQuote {index}:")
            print("Text:", item["Quote"])
            print("Author:", item["Author"])
            print("Category:", item["Category"])

        print("\n")
        print("Total Quotes Extracted:", len(quotes_data))
        print("Output Folder:", output_folder)

    except Exception as e:
        print("Unexpected Error:", e)

html_content = """
<div class="quote">
    <a title="view quote">
        Believe you can and you're halfway there.
    </a>
    <a title="view author">Theodore Roosevelt</a>
</div>

<div class="quote">
    <a title="view quote">
        It always seems impossible until it's done.
    </a>
    <a title="view author">Nelson Mandela</a>
</div>

<div class="quote">
    <a title="view quote">
        The secret of getting ahead is getting started.
    </a>
    <a title="view author">Mark Twain</a>
</div>
"""

scrape_motivational_quotes(html_content)
```
